```python
# ==============================================================================
# PREGUNTA 3: EVALUACIÓN Y MONITOREO DE FAUNA (EJECUCIÓN DE UNA SOLA CORRIDA)
# ==============================================================================

import os
import pandas as pd
from ultralytics import YOLO

# ------------------------------------------------------------------------------
# 1. VERIFICACIÓN Y CARGA DEL MODELO
# ------------------------------------------------------------------------------
print("=== 1. VERIFICACIÓN DEL MODELO ===")
peso_path = 'runs/yolov8s_cct/weights/best.pt'
if os.path.exists(peso_path):
    print(f"Pesos encontrados en: {peso_path}")
    model = YOLO(peso_path)
else:
    print(f"Aviso: No se encontro el archivo de pesos en {peso_path}")

# ------------------------------------------------------------------------------
# 2. MÉTRICAS GLOBALES DE RENDIMIENTO
# ------------------------------------------------------------------------------
print("\n=== 2. MÉTRICAS GLOBALES DE RENDIMIENTO ===")
globales_path = 'eval/globales.csv'
if os.path.exists(globales_path):
    df_global = pd.read_csv(globales_path)
    print(df_global.round(3).to_string(index=False))
else:
    print("Aviso: No se encontro eval/globales.csv")

# ------------------------------------------------------------------------------
# 3. BRECHA DE RENDIMIENTO (CÁMARAS VISTAS VS. NUEVAS)
# ------------------------------------------------------------------------------
print("\n=== 3. BRECHA DE RENDIMIENTO POR ESPECIE (AP50) ===")
brecha_path = 'eval/brecha_por_clase.csv'
if os.path.exists(brecha_path):
    df_brecha = pd.read_csv(brecha_path, index_col=0)
    cols = ['cajas_cis', 'AP50_cis', 'cajas_trans', 'AP50_trans', 'caida_AP50', 'interpretable']
    print(df_brecha[cols].round(3).to_string())
else:
    print("Aviso: No se encontro eval/brecha_por_clase.csv")

# ------------------------------------------------------------------------------
# 4. MONITOREO ECOLÓGICO (CÁMARA 46)
# ------------------------------------------------------------------------------
print("\n=== 4. MONITOREO ECOLÓGICO Y DINÁMICA DE FAUNA (CÁMARA 46) ===")
archivos_monitoreo = {
    "Resumen General": "monitoreo/resumen_cam46.csv",
    "Esfuerzo de Muestreo": "monitoreo/esfuerzo_cam46.csv",
    "Detecciones por Especie": "monitoreo/por_especie_cam46.csv",
    "Comparacion con Etiquetas Reales": "monitoreo/comparacion_cam46.csv"
}

for titulo, ruta in archivos_monitoreo.items():
    if os.path.exists(ruta):
        print(f"\n--- {titulo} ---")
        df = pd.read_csv(ruta, index_col=0)
        print(df.to_string())
    else:
        print(f"\nAviso: No se encontro el archivo {ruta}")

# ------------------------------------------------------------------------------
# 5. ANÁLISIS DE ROBUSTEZ Y DEPENDENCIA DEL FONDO
# ------------------------------------------------------------------------------
print("\n=== 5. ANÁLISIS DE ROBUSTEZ Y DEPENDENCIAS ===")
if os.path.exists('robustez/tamano.csv'):
    print("\n--- Robustez por Tamano de Caja ---")
    df_tam = pd.read_csv('robustez/tamano.csv')
    print(df_tam.to_string(index=False))

if os.path.exists('robustez/oclusion.csv'):
    print("\n--- Robustez por Oclusion ---")
    df_ocl = pd.read_csv('robustez/oclusion.csv')
    print(df_ocl.to_string(index=False))

if os.path.exists('fondo/resultados.csv'):
    print("\n--- Dependencia del Fondo de la Escena ---")
    df_fondo = pd.read_csv('fondo/resultados.csv')
    print(df_fondo.pivot(index='condicion', columns='split', values='pct').to_string())

print("\n==============================================================================")
print("EVALUACIÓN Y AUDITORÍA COMPLETADAS EXITOSAMENTE")
print("==============================================================================")
```
